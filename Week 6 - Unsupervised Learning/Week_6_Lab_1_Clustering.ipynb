### Step 1: Import Dependencies
We need:

````python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
````

### Step 5: Visualize 3 Clusters
We plot the points again, but color them according to the cluster the machine assigned.
We also mark the **Centroids** with a black 'x'.

````python
color = ["red", "pink", "orange", "gray"]

fig, ax1 = plt.subplots(1)

for i in range(n_clusters):
    ax1.scatter(X[y_pred1==i, 0], X[y_pred1==i, 1], 
                marker='o', 
                s=8, 
                c=color[i])

ax1.scatter(centroid1[:,0], centroid1[:,1], marker="x", s=50, c="black")
plt.show()
````

### Step 6: Fix the K Value
The previous result looked okay, but we can see two distinct groups might have been merged. 
Let's try again with `n_clusters=4`.

````python
n_clusters = 4
cluster2 = KMeans(n_clusters=n_clusters, random_state=0)
cluster2.fit(X)

y_pred2 = cluster2.labels_
centroid2 = cluster2.cluster_centers_

# Plot
fig, ax1 = plt.subplots(1)

for i in range(n_clusters):
    ax1.scatter(X[y_pred2==i, 0], X[y_pred2==i, 1], 
                marker='o', 
                s=8, 
                c=color[i])

ax1.scatter(centroid2[:,0], centroid2[:,1], marker="x", s=50, c="black")
plt.show()
````

### Conclusion
With K=4, the clustering perfectly matches the 4 blobs we generated. This demonstrates how K-Means can automatically segregate data.
