```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Model Structure
model = Sequential([
    Dense(64, activation='relu', input_shape=(10,)), # Adjust input shape
    Dense(32, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
# model.fit(X_train, y_train, epochs=10)
```

## Part E: Autoencoder (Unsupervised) IDS


```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

input_dim = 10 # Adjust
input_layer = Input(shape=(input_dim,))
encoder = Dense(32, activation='relu')(input_layer)
code = Dense(8, activation='relu')(encoder)
decoder = Dense(32, activation='relu')(code)
output_layer = Dense(input_dim, activation='sigmoid')(decoder)

autoencoder = Model(input_layer, output_layer)
autoencoder.compile(optimizer='adam', loss='mse')

# Train ONLY on benign data
# autoencoder.fit(X_benign, X_benign, epochs=10)
```
