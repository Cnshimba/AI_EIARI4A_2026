# Week 10 Lab 1: Local LLM Inference on CPU (Quantized)

**Goal:** Run a powerful Large Language Model (Llama 3.1 8B) on a standard laptop (CPU only) using Quantization.

## 1. Why CPU Inference?
Running modern LLMs usually requires massive GPUs. However, by using **Quantization** (reducing precision from 16-bit to 4-bit), we can shrink a 16GB model down to ~5GB, making it runnable on standard RAM with decent speed using optimized libraries like `ctransformers`.

We need the `ctransformers` library, which allows us to run GGUF models efficiently on CPUs.

## 3. Download the Model
We need the **Llama-3.1-8B-Instruct-Q4_K_M.gguf** file.

1.  Download it from Hugging Face (e.g., [Bartowski's repo](https://huggingface.co/bartowski/Meta-Llama-3.1-8B-Instruct-GGUF/blob/main/Meta-Llama-3.1-8B-Instruct-Q4_K_M.gguf)).
2.  Place the `.gguf` file in the same directory as this notebook.

*Note: Ensure you have approx 5GB of free space.*

## 4. Load the Model
We will load the model into RAM. This step might take a few seconds.

```python
from ctransformers import AutoModelForCausalLM

# Update with your exact filename if different
MODEL_PATH = "Meta-Llama-3.1-8B-Instruct-Q4_K_M.gguf"
MODEL_TYPE = "llama"

print(f"Loading {MODEL_PATH} into RAM...")

try:
    # Load the model with gpu_layers=0 to force CPU usage
    llm = AutoModelForCausalLM.from_pretrained(
        MODEL_PATH,
        model_type=MODEL_TYPE,
        gpu_layers=0,
        context_length=2048
    )
    print("Model Loaded Successfully!")
except OSError:
    print("Error: Model file not found. Please ensure the .gguf file is in this folder.")
```

## 5. Chat with the AI
Run the cell below to start a simple chat loop. Type 'exit' to stop.

```python
def chat_with_model():
    print("Starting Chat... (Type 'exit' to quit)")
    while True:
        user_input = input("You: ")
        if user_input.lower() in ["exit", "quit"]:
            break
        
        # Simple Prompt Template for Llama 3 Instruct
        prompt = f"<|begin_of_text|><|start_header_id|>user<|end_header_id|>\n\n{user_input}<|eot_id|><|start_header_id|>assistant<|end_header_id|>\n\n"
        
        print("\nAI: ", end="", flush=True)
        
        # Stream the response
        for text in llm(prompt, stream=True, max_new_tokens=512):
            print(text, end="", flush=True)
        print("\n" + "-"*50)

# Run the chat
if 'llm' in locals():
    chat_with_model()
```
